```python
import os
import pandas as pd
import numpy as np
```

```python
# Define file paths

nykaa_path = os.path.join(
    "Output",
    "Nykaa_Feature_Engineered.csv"
)

purplle_path = os.path.join(
    "Output",
    "Purplle_Feature_Engineered.csv"
)

tira_path = os.path.join(
    "Output",
    "Tira_Feature_Engineered.csv"
)

# Read the three feature-engineered datasets

df_nykaa = pd.read_csv(nykaa_path)
df_purplle = pd.read_csv(purplle_path)
df_tira = pd.read_csv(tira_path)

print("All three feature-engineered datasets loaded successfully!")
```

```python
# Check individual dataset shapes

print("Nykaa shape   :", df_nykaa.shape)
print("Purplle shape :", df_purplle.shape)
print("Tira shape    :", df_tira.shape)

print("\nExpected total rows after consolidation:",
      len(df_nykaa) + len(df_purplle) + len(df_tira))
```

```python
# Validate that all three datasets have the same columns

print("Nykaa vs Purplle columns:",
      df_nykaa.columns.equals(df_purplle.columns))

print("Nykaa vs Tira columns:",
      df_nykaa.columns.equals(df_tira.columns))

print("Purplle vs Tira columns:",
      df_purplle.columns.equals(df_tira.columns))

print("\nNumber of columns in each dataset:")
print("Nykaa   :", len(df_nykaa.columns))
print("Purplle :", len(df_purplle.columns))
print("Tira    :", len(df_tira.columns))
```

```python
# Validate required project columns

required_columns = [
    "Campaign_ID",
    "Campaign_Type",
    "Target_Audience",
    "Duration",
    "Channel_Used",
    "Impressions",
    "Clicks",
    "Leads",
    "Conversions",
    "Revenue",
    "Acquisition_Cost",
    "Language",
    "Engagement_Score",
    "Customer_Segment",
    "Date",
    "Company_Name",
    "Profit",
    "Profit_Flag",
    "Calculated_ROI",
    "Channel_Google",
    "Channel_WhatsApp",
    "Channel_YouTube",
    "Channel_Email",
    "Channel_Instagram",
    "Channel_Facebook"
]

for name, df in {
    "Nykaa": df_nykaa,
    "Purplle": df_purplle,
    "Tira": df_tira
}.items():
    missing_columns = [
        col for col in required_columns
        if col not in df.columns
    ]

    print(f"{name} missing required columns:", missing_columns)

print("\nRequired column validation completed.")
```

```python
# Combine all three feature-engineered datasets

df_combined = pd.concat(
    [df_nykaa, df_purplle, df_tira],
    ignore_index=True
)

print("Combined dataset shape:", df_combined.shape)
```

```python
# Remove the original ROI column

# The original ROI is not used for this project.
# Calculated_ROI is the project ROI metric.

if "ROI" in df_combined.columns:
    df_combined.drop(columns=["ROI"], inplace=True)

print("Original ROI column removed:", "ROI" not in df_combined.columns)
print("Calculated_ROI retained:", "Calculated_ROI" in df_combined.columns)
print("Current shape:", df_combined.shape)
```

```python
# Validate company distribution

company_counts = (
    df_combined["Company_Name"]
    .value_counts()
)

print("Company distribution:")
display(company_counts)

print("\nTotal consolidated rows:", company_counts.sum())
```

```python
# Validate critical project fields

print("Missing Revenue:",
      df_combined["Revenue"].isna().sum())

print("Missing Acquisition Cost:",
      df_combined["Acquisition_Cost"].isna().sum())

print("Missing Profit:",
      df_combined["Profit"].isna().sum())

print("Missing Calculated ROI:",
      df_combined["Calculated_ROI"].isna().sum())

print("\nProfit_Flag distribution:")
display(
    df_combined["Profit_Flag"].value_counts()
)

print("\nDate missing values (intentionally retained):",
      df_combined["Date"].isna().sum())

print("Campaign_ID missing values (intentionally retained):",
      df_combined["Campaign_ID"].isna().sum())
```

```python
# Validate channel encoding

channel_columns = [
    "Channel_Google",
    "Channel_WhatsApp",
    "Channel_YouTube",
    "Channel_Email",
    "Channel_Instagram",
    "Channel_Facebook"
]

print("Channel indicator totals:")
display(
    df_combined[channel_columns]
    .sum()
    .sort_values(ascending=False)
)

channel_indicator_total = (
    df_combined[channel_columns].sum(axis=1)
)

print("Campaigns with no encoded channel:",
      (channel_indicator_total == 0).sum())

print("Campaigns with one or more encoded channels:",
      (channel_indicator_total >= 1).sum())
```

```python
# Final consolidation integrity check

print("Duplicate complete rows:",
      df_combined.duplicated().sum())

print("Number of rows:",
      len(df_combined))

print("Number of columns:",
      len(df_combined.columns))

print("\nFinal column names:")
print(df_combined.columns.tolist())
```

```python
# Save consolidated master dataset

master_output_path = os.path.join(
    "Output",
    "Marketing_Campaign_Feature_Engineered_Master_v2.csv"
)

df_combined.to_csv(
    master_output_path,
    index=False
)

print("Consolidated master dataset saved successfully!")
print("File:", master_output_path)
print("Shape:", df_combined.shape)
```
